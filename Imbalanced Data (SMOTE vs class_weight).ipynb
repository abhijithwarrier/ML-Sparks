**PYTHON SCRIPT TO _HANDLE IMBALANCED CLASSIFICATION USING SMOTE AND CLASS WEIGHTS_. ⚡️⚖️🤖**

This script demonstrates two common strategies for dealing with **imbalanced datasets**:

- **SMOTE** → oversamples the minority class by generating synthetic examples
- **class_weight** → makes the model pay more attention to the minority class during training

This is especially useful in real-world tasks like fraud detection, churn prediction, and rare-event classification.

---

## **📦 Import Required Libraries**

We use scikit-learn for modeling and evaluation, and imbalanced-learn for SMOTE.

```python
import pandas as pd

from sklearn.datasets import make_classification
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import classification_report, confusion_matrix

from imblearn.over_sampling import SMOTE
```

---

## **🧩 Create an Imbalanced Dataset**

We simulate a binary classification problem with strong class imbalance.

```python
X, y = make_classification(
    n_samples=5000,
    n_features=10,
    n_informative=6,
    n_redundant=2,
    weights=[0.90, 0.10],
    random_state=42
)

X = pd.DataFrame(X, columns=[f"feature_{i}" for i in range(10)])
y = pd.Series(y, name="target")
```

This mirrors production problems where one class is much rarer than the other.

---

## **✂️ Split Data into Train and Test Sets**

We keep a separate test set for fair evaluation.

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.3,
    random_state=42,
    stratify=y
)
```

---

## **🚨 Train Baseline Model (No Imbalance Handling)**

We first train a normal classifier without any imbalance strategy.

```python
baseline_model = LogisticRegression(max_iter=5000)
baseline_model.fit(X_train, y_train)

baseline_pred = baseline_model.predict(X_test)

print("=== Baseline Model ===")
print(classification_report(y_test, baseline_pred))
print(confusion_matrix(y_test, baseline_pred))
```

This gives us a reference point.

---

## **🔁 Approach 1 – Handle Imbalance with SMOTE**

SMOTE balances the training data by generating synthetic minority samples.

```python
smote = SMOTE(random_state=42)
X_train_smote, y_train_smote = smote.fit_resample(X_train, y_train)
```

Now we train the same model on the resampled data.

```python
smote_model = LogisticRegression(max_iter=5000)
smote_model.fit(X_train_smote, y_train_smote)

smote_pred = smote_model.predict(X_test)

print("=== SMOTE Model ===")
print(classification_report(y_test, smote_pred))
print(confusion_matrix(y_test, smote_pred))
```

---

## **⚖️ Approach 2 – Handle Imbalance with class_weight**

Instead of resampling, we tell the model to penalize minority-class errors more heavily.

```python
weighted_model = LogisticRegression(
    max_iter=5000,
    class_weight="balanced"
)

weighted_model.fit(X_train, y_train)

weighted_pred = weighted_model.predict(X_test)

print("=== class_weight Model ===")
print(classification_report(y_test, weighted_pred))
print(confusion_matrix(y_test, weighted_pred))
```

This approach keeps the original data unchanged but shifts model learning priorities.

---

## **🔍 Compare the Two Strategies**

- **SMOTE**
    - creates synthetic minority examples
    - can improve minority-class recall
    - changes the training distribution
- **class_weight**
    - leaves data unchanged
    - adjusts how errors are penalized
    - is simpler and often easier to deploy

Both are valid, but the better choice depends on the problem and model type.

---

## **🧠 Key Takeaways**

1. Imbalanced data can make models biased toward the majority class.
2. SMOTE increases minority samples by generating synthetic data.
3. `class_weight="balanced"` changes the loss without resampling data.
4. Both methods often improve minority-class recall.
5. Choosing the right strategy depends on the business objective and model behavior.

---

## **Conclusion**

Handling class imbalance is essential in many real-world ML systems. Techniques like **SMOTE** and **class_weight** help models focus on rare but important classes, making predictions more useful and fair in practice.
