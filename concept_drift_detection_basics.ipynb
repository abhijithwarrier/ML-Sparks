**PYTHON SCRIPT TO _DETECT EARLY SIGNS OF CONCEPT DRIFT USING MODEL PERFORMANCE TRENDS_. ⚡️📉🧠**

This script demonstrates a simple concept drift detection workflow by monitoring model accuracy over time and triggering alerts when performance degrades beyond a defined threshold.

Concept drift is one of the most common reasons production ML models fail after deployment.

---

## **📦 Import Required Libraries**

```python
import pandas as pd
import matplotlib.pyplot as plt
```

---

## **📝 Create Historical Performance Data**

Simulate monthly model performance metrics.

```python
performance_data = pd.DataFrame({
    "month": [
        "Jan", "Feb", "Mar", "Apr",
        "May", "Jun", "Jul", "Aug",
        "Sep", "Oct"
    ],

    "accuracy": [
        0.96, 0.95, 0.95, 0.94,
        0.93, 0.92, 0.90, 0.88,
        0.86, 0.84
    ]
})
```

---

## **🔍 View Monitoring Data**

```python
print(performance_data)
```

---

## **📈 Visualise Accuracy Trend**

```python
plt.figure(figsize=(8, 4))

plt.plot(
    performance_data["month"],
    performance_data["accuracy"],
    marker="o"
)

plt.title("Model Accuracy Trend")
plt.xlabel("Month")
plt.ylabel("Accuracy")

plt.grid(True)
plt.tight_layout()
plt.show()
```

---

## **📊 Calculate Performance Degradation**

```python
baseline_accuracy = performance_data[
    "accuracy"
].iloc[0]

latest_accuracy = performance_data[
    "accuracy"
].iloc[-1]

accuracy_drop = (
    baseline_accuracy -
    latest_accuracy
)

print(
    "Accuracy Drop:",
    round(accuracy_drop, 4)
)
```

---

## **🚨 Simple Drift Detection Rule**

```python
DRIFT_THRESHOLD = 0.10

if accuracy_drop > DRIFT_THRESHOLD:
    print(
        "⚠️ Concept Drift Detected"
    )
else:
    print(
        "✅ No Significant Drift"
    )
```

---

## **📈 Rolling Accuracy Monitoring**

Rolling averages smooth temporary fluctuations.

```python
performance_data[
    "rolling_accuracy"
] = (
    performance_data["accuracy"]
    .rolling(window=3)
    .mean()
)

print(
    performance_data
)
```

---

## **📊 Plot Rolling Trend**

```python
plt.figure(figsize=(8, 4))

plt.plot(
    performance_data["month"],
    performance_data["accuracy"],
    marker="o",
    label="Actual Accuracy"
)

plt.plot(
    performance_data["month"],
    performance_data["rolling_accuracy"],
    marker="s",
    label="Rolling Average"
)

plt.title(
    "Concept Drift Monitoring"
)

plt.xlabel("Month")
plt.ylabel("Accuracy")

plt.legend()

plt.grid(True)
plt.tight_layout()
plt.show()
```

---

## **🔍 Why Concept Drift Happens**

Concept drift occurs when:

- customer behaviour changes
- business rules evolve
- economic conditions shift
- fraud patterns adapt
- user preferences change

The relationship between inputs and outputs no longer matches what the model learned during training.

---

## **🧠 Real-World Example**

A fraud model trained in 2024 may learn: _Transaction Pattern → Fraud Risk_

By 2026, fraudsters may change behaviour completely.

The model still uses old patterns, causing performance degradation.

---

## **🧠 Key Takeaways**

1. Concept drift occurs when feature-target relationships change over time.
2. Performance monitoring is often the first signal of drift.
3. Accuracy trends can reveal gradual degradation.
4. Rolling averages help identify long-term patterns.
5. Drift detection is a critical ML Systems practice.

---

## **Conclusion**

Machine learning models operate in environments that continuously evolve. Monitoring performance trends and detecting concept drift early helps prevent silent model failures and ensures reliable predictions over time.
