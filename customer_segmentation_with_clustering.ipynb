**PYTHON SCRIPT TO _SEGMENT CUSTOMERS INTO MEANINGFUL GROUPS USING UNSUPERVISED LEARNING (KMEANS CLUSTERING)_. ⚡📊🧠**

This script demonstrates how businesses group customers based on behavior or attributes using **KMeans clustering**. Customer segmentation helps marketing, product, and growth teams tailor strategies for different customer groups.

---

## **📦 Import Required Libraries**

**We use NumPy, pandas, scikit-learn, and matplotlib for clustering and visualization.**

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler
```

---

## **🧩 Create a Customer Dataset**

**We simulate customer data with spending behavior and income.**

```python
np.random.seed(42)

data = {
    "annual_income": np.random.randint(20000, 120000, 300),
    "spending_score": np.random.randint(1, 100, 300)
}

df = pd.DataFrame(data)
```

This mirrors common customer analytics features used in retail and e-commerce.

---

## **📏 Feature Scaling**

**Scaling is essential for distance-based clustering algorithms like KMeans.**

```python
scaler = StandardScaler()
X_scaled = scaler.fit_transform(df)
```

---

## **🔍 Apply KMeans Clustering**

**We segment customers into 4 groups.**

```python
kmeans = KMeans(
    n_clusters=4,
    random_state=42
)

df["cluster"] = kmeans.fit_predict(X_scaled)
```

---

## **📊 Visualize Customer Segments**

**Plot customer groups based on income and spending behavior.**

```python
plt.figure(figsize=(8, 5))

for cluster in df["cluster"].unique():
    subset = df[df["cluster"] == cluster]
    plt.scatter(
        subset["annual_income"],
        subset["spending_score"],
        label=f"Cluster {cluster}"
    )

plt.xlabel("Annual Income")
plt.ylabel("Spending Score")
plt.title("Customer Segmentation using KMeans")
plt.legend()
plt.grid(True)
plt.show()
```

---

## **🧪 Interpret the Segments**

Each cluster typically represents:

- High income – high spenders
- High income – low spenders
- Low income – high spenders
- Low income – low spenders

These insights guide **targeted marketing and personalization**.
