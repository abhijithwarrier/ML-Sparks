**PYTHON SCRIPT TO IDENTIFY AND PREVENT DATA LEAKAGE IN MACHINE LEARNING PIPELINES. ⚡️🚨📊**

This script demonstrates how **data leakage** can silently inflate model performance and how to detect and fix it using proper ML workflows.

---

## **📦 Import Required Libraries**

```python
import pandas as pd

from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import train_test_split, cross_val_score
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score
```

---

## **🧩 Load Dataset**

```python
data = load_breast_cancer()

X = pd.DataFrame(data.data, columns=data.feature_names)
y = data.target
```

---

## **🚨 ❌ Wrong Approach – Data Leakage**

Here we apply preprocessing **before splitting**, causing leakage.

```python
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)  # ❌ leakage

X_train_leak, X_test_leak, y_train_leak, y_test_leak = train_test_split(
    X_scaled,
    y,
    test_size=0.3,
    random_state=42,
    stratify=y
)

model_leak = LogisticRegression(max_iter=5000)
model_leak.fit(X_train_leak, y_train_leak)

y_pred_leak = model_leak.predict(X_test_leak)

print("❌ Accuracy with Leakage:", accuracy_score(y_test_leak, y_pred_leak))
```

---

## **🔍 Why This Is Wrong**

- The scaler learned from **entire dataset**
- Test data influenced training
- Accuracy appears artificially high

---

## **✅ Correct Approach – No Leakage**

Split first, then apply transformations.

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.3,
    random_state=42,
    stratify=y
)

scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

model = LogisticRegression(max_iter=5000)
model.fit(X_train_scaled, y_train)

y_pred = model.predict(X_test_scaled)

print("✅ Accuracy without Leakage:", accuracy_score(y_test, y_pred))
```

---

## **🚀 Best Practice – Using Pipeline**

Pipelines ensure transformations are applied correctly.

```python
pipeline = Pipeline([
    ("scaler", StandardScaler()),
    ("model", LogisticRegression(max_iter=5000))
])

scores = cross_val_score(pipeline, X, y, cv=5)

print("🚀 Cross-validated Accuracy:", scores.mean())
```

---

## **🔍 How to Detect Data Leakage**

Signs of leakage:

- unusually high accuracy
- model performs poorly in real-world testing
- large drop between train and production performance

---

## **🧠 Key Takeaways**

1. Data leakage occurs when test data influences training.
2. Preprocessing before splitting is a common mistake.
3. Leakage leads to unrealistic performance.
4. Always split data before transformations.
5. Pipelines help prevent leakage automatically.

---

## **Conclusion**

Data leakage is one of the most dangerous issues in machine learning. It can make models appear highly accurate while failing in real-world scenarios. By following proper workflows and using pipelines, we can build reliable and trustworthy ML systems.
