PYTHON SCRIPT TO *REDUCE HIGH-DIMENSIONAL DATA USING PRINCIPAL COMPONENT ANALYSIS (PCA)*. ⚡📉🤖

This script demonstrates how **Principal Component Analysis (PCA)** reduces high-dimensional data into fewer components while preserving maximum variance. PCA is widely used for visualization, noise reduction, and speeding up downstream ML models.

---

## **📦 Import Required Libraries**

**We import sklearn utilities, NumPy, and matplotlib for PCA and visualization.**

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import load_wine
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
```

---

## **🧩 Load a High-Dimensional Dataset**

**We use the Wine dataset, which contains multiple numerical features.**

```python
data = load_wine()
X = data.data
y = data.target
```

---

## **📏 Standardize the Features**

**PCA is sensitive to scale, so we standardize the data first.**

```python
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
```

---

## **📉 Apply PCA for Dimensionality Reduction**

**Reduce the data to two principal components.**

```python
pca = PCA(n_components=2)
X_pca = pca.fit_transform(X_scaled)
```

---

## **📊 Visualize the Reduced Data**

**Plot the data in 2D using the principal components.**

```python
plt.figure(figsize=(8, 5))
plt.scatter(X_pca[:, 0], X_pca[:, 1], c=y, cmap="viridis")
plt.xlabel("Principal Component 1")
plt.ylabel("Principal Component 2")
plt.title("PCA – 2D Projection of Wine Dataset")
plt.grid(True)
plt.show()
```

This visualization reveals how well classes separate in reduced dimensions.

---

## **📈 Explained Variance Ratio**

**Check how much variance each principal component captures.**

```python
print("Explained variance ratio:", pca.explained_variance_ratio_)
print("Total variance captured:", np.sum(pca.explained_variance_ratio_))
```

---

## **🎯 Why PCA Is Useful**

- Reduces dimensionality without labels
- Helps visualize complex datasets
- Removes correlated and noisy features
- Speeds up ML models
- Improves interpretability

---

## **Key Takeaways**

1. PCA reduces high-dimensional data while preserving variance.
2. Feature scaling is critical before applying PCA.
3. Principal components are linear combinations of original features.
4. Explained variance shows how much information is retained.
5. PCA is widely used before clustering and visualization tasks.
