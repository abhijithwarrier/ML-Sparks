**PYTHON SCRIPT TO *FIND THE OPTIMAL NUMBER OF CLUSTERS USING ELBOW METHOD & SILHOUETTE SCORE*. ⚡📊🤖**

This script demonstrates two popular techniques used with **KMeans clustering** to decide the optimal number of clusters: the **Elbow Method** and the **Silhouette Score**. These methods help evaluate clustering quality before finalizing k.

---

## **📦 Import Required Libraries**

**We import sklearn utilities, NumPy, and matplotlib for clustering evaluation.**

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import make_blobs
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score
```

---

## **🧩 Generate Unlabeled Data**

**Create a synthetic dataset with natural cluster structure.**

```python
X, _ = make_blobs(
    n_samples=400,
    centers=4,
    cluster_std=0.60,
    random_state=42
)
```

---

## **📉 Elbow Method – Inertia vs Number of Clusters**

**Compute inertia (within-cluster sum of squares) for different values of k.**

```python
inertias = []
k_range = range(2, 11)

for k in k_range:
    kmeans = KMeans(n_clusters=k, random_state=42)
    kmeans.fit(X)
    inertias.append(kmeans.inertia_)
```

---

## **📊 Plot the Elbow Curve**

**Visualize how inertia decreases as k increases.**

```python
plt.figure(figsize=(8, 5))
plt.plot(k_range, inertias, marker="o")
plt.xlabel("Number of Clusters (k)")
plt.ylabel("Inertia")
plt.title("Elbow Method for Optimal k")
plt.grid(True)
plt.show()
```

The “elbow” point indicates a good balance between simplicity and fit.

---

## **📐 Silhouette Score – Cluster Quality Metric**

**Calculate silhouette scores for each k to measure cluster separation.**

```python
silhouette_scores = []

for k in k_range:
    kmeans = KMeans(n_clusters=k, random_state=42)
    labels = kmeans.fit_predict(X)
    score = silhouette_score(X, labels)
    silhouette_scores.append(score)
```

---

## **📊 Plot Silhouette Scores**

**Higher silhouette score indicates better-defined clusters.**

```python
plt.figure(figsize=(8, 5))
plt.plot(k_range, silhouette_scores, marker="o", color="green")
plt.xlabel("Number of Clusters (k)")
plt.ylabel("Silhouette Score")
plt.title("Silhouette Score vs Number of Clusters")
plt.grid(True)
plt.show()
```

---

## **🎯 How to Interpret the Results**

- **Elbow Method:** Look for a sharp bend where inertia reduction slows
- **Silhouette Score:** Choose k with the **highest score**
- Use **both together** for more reliable cluster selection

---

## ✅ **Key Takeaways**

1. Choosing the right number of clusters is critical in unsupervised learning.
2. The Elbow Method highlights diminishing returns as k increases.
3. Silhouette Score measures how well clusters are separated.
4. Combining both methods gives more confident decisions.
5. These techniques are essential before finalizing KMeans models.
