**PYTHON SCRIPT TO ENCODE HIGH-CARDINALITY CATEGORICAL FEATURES FOR SCALABLE MACHINE LEARNING. ⚡️🏷️📊**

This script demonstrates how to handle categorical columns with many unique values (high cardinality) using practical encoding strategies suitable for real-world tabular ML systems.

---

## **📦 Import Required Libraries**

```python
import numpy as np
import pandas as pd

from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score
```

---

## **🧩 Create Sample Dataset**

We simulate a dataset containing a high-cardinality categorical feature.

```python
np.random.seed(42)

n_samples = 1000

data = pd.DataFrame({
    "user_id": [f"user_{i}" for i in range(n_samples)],

    "city": np.random.choice(
        [f"city_{i}" for i in range(200)],
        size=n_samples
    ),

    "age": np.random.randint(18, 60, size=n_samples),

    "purchased": np.random.randint(0, 2, size=n_samples)
})
```

The `city` column contains many unique categories.

---

## **🚨 Problem with One-Hot Encoding**

Applying one-hot encoding directly:

```python
pd.get_dummies(data["city"])
```

can create:

- hundreds or thousands of columns
- sparse matrices
- large memory usage
- slower training pipelines

---

## **✅ Frequency Encoding**

A practical alternative is frequency encoding.

```python
city_freq = data["city"].value_counts()

data["city_freq_encoded"] = data["city"].map(city_freq)
```

Each category is replaced by its occurrence frequency.

---

## **✂️ Prepare Features**

```python
X = data[["age", "city_freq_encoded"]]
y = data["purchased"]
```

---

## **✂️ Split Data**

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.3,
    random_state=42
)
```

---

## **🤖 Train Model**

```python
model = LogisticRegression(max_iter=5000)

model.fit(X_train, y_train)
```

---

## **📊 Evaluate Model**

```python
y_pred = model.predict(X_test)

print("Accuracy:", accuracy_score(y_test, y_pred))
```

---

## **🔍 Why High Cardinality Encoding Matters**

High-cardinality features are common in:

- user IDs
- product IDs
- city names
- transaction categories
- search queries

Naive encoding methods can explode feature dimensions and reduce scalability.

---

## **🧠 Other Common Techniques**

Besides frequency encoding:

- Target Encoding
- Hash Encoding
- Leave-One-Out Encoding
- Embedding Layers (deep learning)

are also widely used for categorical handling.

---

## **🧠 Key Takeaways**

1. High-cardinality features contain many unique categories.
2. One-hot encoding becomes inefficient at scale.
3. Frequency encoding is a compact alternative.
4. Efficient encoding improves scalability and training speed.
5. Proper categorical encoding is critical in tabular ML.

---

## **Conclusion**

Handling high-cardinality categorical data is an important challenge in machine learning systems. By using practical techniques like frequency encoding, we can build scalable and efficient tabular ML pipelines without exploding feature dimensions.
