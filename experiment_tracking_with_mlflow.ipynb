## 🚀 **Why MLflow?**

Without experiment tracking:

- spreadsheets for results
- forgotten hyperparameters
- difficult model comparisons
- poor reproducibility

With MLflow:

- centralised experiment history
- automatic metric logging
- model artifact management
- easier collaboration
- reproducible experiments

## 🧠 **Key Takeaways**

1. MLflow records every machine learning experiment.
2. Parameters, metrics, and models are stored together.
3. Experiment tracking improves reproducibility.
4. Comparing multiple runs helps identify the best model.
5. MLflow is one of the most widely used open-source MLOps platforms.
