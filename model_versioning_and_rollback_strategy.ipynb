**PYTHON SCRIPT TO SIMULATE MODEL VERSIONING AND AUTOMATIC ROLLBACK IN A PRODUCTION ML SYSTEM. ⚡️📦🔄**

Training a machine learning model is only the beginning. As models are improved over time, organizations need a reliable way to manage multiple versions and safely recover if a newly deployed model performs poorly.

This is where **Model Versioning** and **Rollback Strategies** become essential.

In this project, we’ll simulate a simple model registry, compare model performance, and automatically decide whether to promote a new version or roll back to the previous stable model.

---

## **📦 Import Required Libraries**

````python
import joblib
import pandas as pd

from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import train_test_split

from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier

from sklearn.metrics import (
    accuracy_score,
    f1_score
)
````

---

## **🧩 Load Dataset**

````python
data = load_breast_cancer()

X = pd.DataFrame(
    data.data,
    columns=data.feature_names
)

y = pd.Series(
    data.target,
    name="target"
)
````

---

## **✂️ Split Dataset**

````python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.30,
    random_state=42,
    stratify=y
)
````

---

## **🤖 Train Version 1 (Production Model)**

````python
model_v1 = LogisticRegression(
    max_iter=5000
)

model_v1.fit(
    X_train,
    y_train
)
````

---

## **🚀 Train Version 2 (Candidate Model)**

````python
model_v2 = RandomForestClassifier(
    n_estimators=200,
    random_state=42
)

model_v2.fit(
    X_train,
    y_train
)
````

---

## **📊 Evaluate Both Versions**

````python
pred_v1 = model_v1.predict(
    X_test
)

pred_v2 = model_v2.predict(
    X_test
)

f1_v1 = f1_score(
    y_test,
    pred_v1
)

f1_v2 = f1_score(
    y_test,
    pred_v2
)

print("Version 1 F1:", round(f1_v1, 4))
print("Version 2 F1:", round(f1_v2, 4))
````

---

## **📦 Create a Simple Model Registry**

````python
model_registry = {

    "v1": {
        "status": "Production",
        "model": model_v1
    },

    "v2": {
        "status": "Candidate",
        "model": model_v2
    }

}
````

---

## **🔄 Deployment Decision**

````python
if f1_v2 >= f1_v1:
    print("🚀 Promote Version 2")
    active_version = "v2"
else:
    print("⛔ Roll Back to Version 1")
    active_version = "v1"
````

---

## **💾 Save Active Model**

````python
joblib.dump(
    model_registry[active_version]["model"],
    "active_model.pkl"
)
````

---

## **📂 Load Active Model**

````python
active_model = joblib.load(
    "active_model.pkl"
)

predictions = active_model.predict(
    X_test.iloc[:5]
)

print(predictions)
````

---

## **📋 Version Summary**

````python
summary = pd.DataFrame({

    "Version": [
        "v1",
        "v2"
    ],

    "Status": [
        model_registry["v1"]["status"],
        model_registry["v2"]["status"]
    ],

    "F1 Score": [
        f1_v1,
        f1_v2
    ]

})

print(summary)
````
