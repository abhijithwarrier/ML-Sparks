**PYTHON SCRIPT TO *BUILD A SIMPLE CONTENT-BASED MOVIE RECOMMENDATION SYSTEM USING COSINE SIMILARITY*. ⚡🎬🧠**

This script demonstrates how recommendation systems work by measuring **similarity between movies** based on their features. We use **cosine similarity** to recommend movies that are most similar to a given movie.

---

## **📦 Import Required Libraries**

**We use pandas and scikit-learn utilities for vectorization and similarity computation.**

```python
import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity
```

---

## **🧩 Create a Simple Movie Dataset**

**We simulate a small movie dataset with genres and descriptions.**

```python
movies = {
    "title": [
        "Inception",
        "The Dark Knight",
        "Interstellar",
        "The Matrix",
        "Avengers: Endgame"
    ],
    "description": [
        "dream manipulation sci-fi thriller",
        "dark gritty superhero action crime",
        "space exploration time sci-fi drama",
        "virtual reality dystopian sci-fi action",
        "superheroes action time travel battle"
    ]
}

df = pd.DataFrame(movies)
```

This mirrors how real systems work with metadata like genres, tags, or plot summaries.

---

## **🔠 Convert Text to TF-IDF Vectors**

**TF-IDF transforms movie descriptions into numeric vectors.**

```python
vectorizer = TfidfVectorizer(stop_words="english")
tfidf_matrix = vectorizer.fit_transform(df["description"])
```

Each movie is now represented as a weighted feature vector.

---

## **📐 Compute Cosine Similarity**

**Cosine similarity measures how close two movies are in vector space.**

```python
cosine_sim = cosine_similarity(tfidf_matrix, tfidf_matrix)
```

Values range from **0 (no similarity)** to **1 (identical)**.

---

## **🎯 Build a Recommendation Function**

**Recommend movies similar to a given title.**

```python
def recommend(movie_title, top_n=3):
    idx = df.index[df["title"] == movie_title][0]
    sim_scores = list(enumerate(cosine_sim[idx]))
    sim_scores = sorted(sim_scores, key=lambda x: x[1], reverse=True)

    recommendations = [
        df["title"][i]
        for i, score in sim_scores[1 : top_n + 1]
    ]

    return recommendations
```

---

## **🎬 Get Movie Recommendations**

**Test the recommendation system.**

```python
print("Recommended movies for Inception:")
print(recommend("Inception"))
```

This returns movies with similar themes and keywords.

---

## **🧪 Why Cosine Similarity Works for Recommendations**

- Independent of document length
- Captures semantic similarity in text
- Scales well to large catalogs
- Foundation of many content-based recommender systems

---

## **Key Takeaways**

1. Content-based recommendation uses item similarity instead of user behavior.
2. TF-IDF converts text metadata into numeric features.
3. Cosine similarity measures closeness between items effectively.
4. Simple recommenders can be built with very little data.
5. This approach scales to real-world recommendation engines.
