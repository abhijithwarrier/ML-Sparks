**PYTHON SCRIPT TO *TEST NEW ML MODELS IN PRODUCTION WITHOUT IMPACTING REAL USERS*. 👥🕶️🤖**

Deploying a new Machine Learning model directly into production always carries some risk. Even if a model performs well during testing, real-world traffic can expose issues that weren't visible during development.

**Shadow Deployment** is a production deployment strategy where a new model receives the same live requests as the production model but **its predictions are never returned to users**. Instead, its outputs are logged and compared with the production model to evaluate performance safely.

In this project, we'll simulate a Shadow Deployment pipeline using Python.

We'll learn how to:

- Understand Shadow Deployment
- Compare production and shadow model predictions
- Log prediction differences
- Evaluate shadow model performance
- Understand when Shadow Deployment should be used

---

## 📦 Import Required Libraries

````python
import random
import pandas as pd
````

We'll use Python to simulate production traffic and compare predictions from production and shadow models.

---

## 🤖 Create the Production Model

````python
def production_model(feature):

    if feature > 0.5:
        return "Approved"

    return "Rejected"
````

This represents the model currently serving predictions to users.

---

## 🆕 Create the Shadow Model

````python
def shadow_model(feature):

    if feature > 0.45:
        return "Approved"

    return "Rejected"
````

The shadow model represents a newer version that we want to evaluate before replacing the production model.

---

## 📊 Simulate Incoming Requests

````python
requests = [
    round(random.random(), 2)
    for _ in range(20)
]
````

Each value represents a feature received from a live production request.

---

## 🚀 Send Requests to Both Models

````python
logs = []

for feature in requests:

    production_prediction = production_model(feature)

    shadow_prediction = shadow_model(feature)

    logs.append({
        "Feature": feature,
        "Production": production_prediction,
        "Shadow": shadow_prediction
    })
````

The production prediction is returned to the user, while the shadow prediction is recorded only for analysis.

---

## 📄 Create a Comparison Table

````python
results = pd.DataFrame(logs)

print(results)
````

This allows us to compare predictions from both models side by side.

---

## ⚠️ Detect Prediction Differences

````python
results["Different"] = (
    results["Production"] !=
    results["Shadow"]
)

print(results)
````

Rows marked as **True** indicate requests where the two models disagree.

---

## 📈 Calculate Agreement Rate

````python
agreement = (
    (~results["Different"]).mean()
) * 100

print(
    f"Agreement Rate: {agreement:.2f}%"
)
````

A high agreement rate indicates that the new model behaves similarly to the production model.

---

## 📊 Count Disagreements

````python
print(
    results["Different"].value_counts()
)
````

This helps quantify how often the new model makes different decisions.

```
        │

        ▼

        │

```

Shadow Deployment allows organisations to evaluate a new model under real production traffic without exposing users to its predictions.

## 🧠 Key Takeaways

- Shadow Deployment evaluates new models using live production traffic.
- Users always receive predictions from the production model.
- Shadow model predictions are logged for comparison only.
- Agreement rates help determine model readiness.
- Shadow Deployment reduces deployment risk before full production rollout.

---

## Conclusion

Shadow Deployment is one of the safest strategies for validating Machine Learning models in production. By allowing a new model to process real user requests without affecting the user experience, organiSations can measure its behaviour, compare it with the existing production model, and identify potential issues before deployment. This approach significantly reduces deployment risk and builds confidence in new model releases.
