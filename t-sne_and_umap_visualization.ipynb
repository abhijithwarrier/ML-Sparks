**PYTHON SCRIPT TO *VISUALIZE HIGH-DIMENSIONAL DATA USING t-SNE AND UMAP*. ⚡📉🧠**

This script demonstrates how to use **t-SNE** and **UMAP** to project high-dimensional data into 2D space for visualization. These techniques are widely used to explore structure, clusters, and patterns that are difficult to observe in raw feature space.

---

## **📦 Import Required Libraries**

**We import sklearn for datasets and t-SNE, and UMAP for advanced manifold visualization.**

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import load_digits
from sklearn.preprocessing import StandardScaler
from sklearn.manifold import TSNE
import umap.umap_ as umap
```

---

## **🧩 Load a High-Dimensional Dataset**

**We use the Digits dataset (64-dimensional images).**

```python
digits = load_digits()
X = digits.data
y = digits.target
```

---

## **📏 Standardize the Data**

**Scaling improves the quality of t-SNE and UMAP projections.**

```python
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
```

---

## **🔍 Apply t-SNE for Visualization**

**t-SNE preserves local neighborhood structure in lower dimensions.**

```python
tsne = TSNE(
    n_components=2,
    perplexity=30,
    random_state=42
)

X_tsne = tsne.fit_transform(X_scaled)
```

---

## **📊 Visualize t-SNE Projection**

**Plot the 2D embedding colored by class labels.**

```python
plt.figure(figsize=(8, 5))
plt.scatter(X_tsne[:, 0], X_tsne[:, 1], c=y, cmap="tab10", s=15)
plt.title("t-SNE Visualization of Digits Dataset")
plt.xlabel("Component 1")
plt.ylabel("Component 2")
plt.show()
```

---

## **🌐 Apply UMAP for Visualization**

**UMAP preserves both local and global data structure efficiently.**

```python
umap_model = umap.UMAP(
    n_components=2,
    n_neighbors=15,
    min_dist=0.1,
    random_state=42
)

X_umap = umap_model.fit_transform(X_scaled)
```

---

## **📊 Visualize UMAP Projection**

**Plot the UMAP embedding for comparison.**

```python
plt.figure(figsize=(8, 5))
plt.scatter(X_umap[:, 0], X_umap[:, 1], c=y, cmap="tab10", s=15)
plt.title("UMAP Visualization of Digits Dataset")
plt.xlabel("Component 1")
plt.ylabel("Component 2")
plt.show()
```

---

## **🧪 When to Use t-SNE vs UMAP**

- **t-SNE:** Excellent for visualizing tight local clusters
- **UMAP:** Faster, scalable, and preserves global structure better
- Both are visualization tools, not feature transformers for training

---

## **Key Takeaways**

1. t-SNE and UMAP reveal structure in high-dimensional data visually.
2. Scaling improves the quality of embeddings.
3. t-SNE focuses on local neighborhood preservation.
4. UMAP balances local and global structure efficiently.
5. These techniques are ideal for exploratory data analysis.
