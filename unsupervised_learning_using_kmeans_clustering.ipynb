**PYTHON SCRIPT TO *PERFORM UNSUPERVISED LEARNING USING KMEANS CLUSTERING*. ⚡🤖📊**

This script demonstrates how to apply **KMeans clustering** to group unlabeled data into distinct clusters.

KMeans is one of the simplest and most effective unsupervised algorithms, widely used in customer segmentation, anomaly detection, and pattern discovery.

---

## **📦 Import Required Libraries**

**We import sklearn, NumPy, and matplotlib for clustering and visualization.**

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import make_blobs
from sklearn.cluster import KMeans
```

---

## **🧩 Generate Synthetic Unlabeled Data**

**We create a dataset with 3 clusters using make_blobs().**

```python
X, y_true = make_blobs(
    n_samples=300,
    centers=3,
    cluster_std=0.60,
    random_state=42
)
```

---

## **🤖 Apply KMeans Clustering**

**We initialize KMeans with 3 clusters and fit it to the data.**

```python
kmeans = KMeans(n_clusters=3, random_state=42)
kmeans.fit(X)
```

---

## **🔮 Retrieve Cluster Assignments**

**KMeans labels each sample with its assigned cluster.**

```python
labels = kmeans.labels_
centroids = kmeans.cluster_centers_
```

---

## **📊 Visualize the Clusters**

**Plot clustered points with their corresponding centroids.**

```python
plt.figure(figsize=(8, 5))
plt.scatter(X[:, 0], X[:, 1], c=labels, s=40, cmap="viridis")
plt.scatter(centroids[:, 0], centroids[:, 1], c="red", s=200, marker="X")
plt.title("KMeans Clustering Result")
plt.xlabel("Feature 1")
plt.ylabel("Feature 2")
plt.grid(True)
plt.show()
```

---

## **🧪 Predict Cluster for New Points**

**KMeans can assign clusters to new unseen data.**

```python
new_points = np.array([[0, 2], [3, 4], [-1, -2]])
preds = kmeans.predict(new_points)
print("Cluster predictions for new points:", preds)
```

---

## **🎯 Summary**

- KMeans separates data based on distances to cluster centroids
- The algorithm iteratively refines centroid positions
- Labels represent **cluster membership**, not true classes
- Useful in segmentation, trend discovery, and grouping patterns
