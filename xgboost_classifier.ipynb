**PYTHON SCRIPT TO *TRAIN A HIGH-PERFORMANCE GRADIENT BOOSTING MODEL USING XGBOOST FOR TABULAR DATA*. ⚡️🚀🤖**

This script demonstrates how to use **XGBoost**, one of the most powerful and widely used boosting algorithms, to build a high-performance classification model for structured data.

---

## **📦 Import Required Libraries**

We use XGBoost along with scikit-learn utilities.

```python
import pandas as pd

from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score, classification_report

from xgboost import XGBClassifier
```

---

## **🧩 Load Dataset**

```python
data = load_breast_cancer()

X = pd.DataFrame(data.data, columns=data.feature_names)
y = data.target
```

---

## **✂️ Split Data**

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.3,
    random_state=42,
    stratify=y
)
```

---

## **🚀 Train XGBoost Classifier**

We initialise and train the model.

```python
xgb_model = XGBClassifier(
    n_estimators=200,
    learning_rate=0.1,
    max_depth=4,
    subsample=0.8,
    colsample_bytree=0.8,
    random_state=42,
    use_label_encoder=False,
    eval_metric="logloss"
)

xgb_model.fit(X_train, y_train)
```

---

## **📊 Evaluate Model Performance**

```python
y_pred = xgb_model.predict(X_test)

print("Accuracy:", accuracy_score(y_test, y_pred))
print("\nClassification Report:\n")
print(classification_report(y_test, y_pred))
```
